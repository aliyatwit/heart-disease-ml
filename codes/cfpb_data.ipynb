```python
# Combine all files into one dataframe
import pandas as pd
import glob
import os

folder = "CCDB"
files = sorted(glob.glob(os.path.join(folder, "CCDB_Export_*.csv")))

dataframes = []

for file in files:
    print(f"Loading: {os.path.basename(file)}")

    try:
        temp_df = pd.read_csv(
            file,
            engine="python",
            on_bad_lines="skip"
        )

        dataframes.append(temp_df)
        print(f"  Loaded {len(temp_df):,} rows")

    except Exception as e:
        print(f"  Failed: {e}")

if dataframes:
    df = pd.concat(dataframes, ignore_index=True, sort=False)
    print(f"\nCombined shape: {df.shape}")
    display(df.head())
else:
    print("No files loaded successfully.")
```

```python
#EDA 

import pandas as pd
import matplotlib.pyplot as plt

# Use your existing cleaned DataFrame
eda_df = df.copy()

eda_df["Date received"] = pd.to_datetime(
    eda_df["Date received"], errors="coerce"
)

# --------------------------------------------------
# FIGURE 1: Monthly complaint volume
# --------------------------------------------------
monthly = (
    eda_df.dropna(subset=["Date received"])
    .set_index("Date received")
    .resample("MS")
    .size()
)

plt.figure(figsize=(14, 5))
plt.plot(monthly.index, monthly.values)
plt.title("Monthly Consumer Complaint Volume")
plt.xlabel("Date")
plt.ylabel("Number of Complaints")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# --------------------------------------------------
# FIGURE 2: Most common financial products
# --------------------------------------------------
top_products = eda_df["Product"].value_counts().head(10).sort_values()

plt.figure(figsize=(11, 6))
top_products.plot(kind="barh")
plt.title("Top 10 Financial Products by Complaint Volume")
plt.xlabel("Number of Complaints")
plt.ylabel("Product")
plt.tight_layout()
plt.show()


# --------------------------------------------------
# FIGURE 3: Most common complaint issues
# --------------------------------------------------
top_issues = eda_df["Issue"].value_counts().head(10).sort_values()

plt.figure(figsize=(11, 6))
top_issues.plot(kind="barh")
plt.title("Top 10 Consumer Complaint Issues")
plt.xlabel("Number of Complaints")
plt.ylabel("Issue")
plt.tight_layout()
plt.show()


# --------------------------------------------------
# FIGURE 4: Company response outcomes
# --------------------------------------------------
response_counts = (
    eda_df["Company response to consumer"]
    .fillna("Missing")
    .value_counts()
    .head(8)
    .sort_values()
)

plt.figure(figsize=(11, 6))
response_counts.plot(kind="barh")
plt.title("Most Common Company Response Outcomes")
plt.xlabel("Number of Complaints")
plt.ylabel("Response Outcome")
plt.tight_layout()
plt.show()


# --------------------------------------------------
# FIGURE 5: Consumer narrative availability
# --------------------------------------------------
narrative_available = eda_df[
    "Consumer complaint narrative"
].notna()

narrative_counts = pd.Series({
    "Narrative available": int(narrative_available.sum()),
    "No narrative": int((~narrative_available).sum())
})

plt.figure(figsize=(7, 5))
narrative_counts.plot(kind="bar")
plt.title("Availability of Consumer Complaint Narratives")
plt.xlabel("Narrative Status")
plt.ylabel("Number of Complaints")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# --------------------------------------------------
# FIGURE 6: Complaints by state
# --------------------------------------------------
top_states = (
    eda_df["State"]
    .dropna()
    .value_counts()
    .head(15)
    .sort_values()
)

plt.figure(figsize=(10, 6))
top_states.plot(kind="barh")
plt.title("Top 15 States by Complaint Volume")
plt.xlabel("Number of Complaints")
plt.ylabel("State")
plt.tight_layout()
plt.show()
```

```python

# Overall dataset statistics
print("DATASET SUMMARY")
print("-" * 45)
print(f"Total complaints: {len(eda_df):,}")
print(f"Unique products: {eda_df['Product'].nunique():,}")
print(f"Unique issues: {eda_df['Issue'].nunique():,}")
print(f"Unique companies: {eda_df['Company'].nunique():,}")
print(f"States represented: {eda_df['State'].nunique():,}")

# Complaints by year
print("\nCOMPLAINTS BY YEAR")
complaints_by_year = (
    eda_df.dropna(subset=["Date received"])
    .groupby(eda_df["Date received"].dt.year)
    .size()
    .rename("Complaint Count")
)
display(complaints_by_year.to_frame())

# Narrative coverage
narrative_pct = (
    eda_df["Consumer complaint narrative"].notna().mean() * 100
)
print(f"\nComplaints with narratives: {narrative_pct:.2f}%")

# Timely response statistics
print("\nTIMELY RESPONSE")
display(
    eda_df["Timely response?"]
    .value_counts(dropna=False)
    .rename_axis("Timely Response")
    .to_frame("Count")
)

# Missing data
print("\nMISSING DATA")
missing = pd.DataFrame({
    "Missing Count": eda_df.isna().sum(),
    "Missing Percentage": (eda_df.isna().mean() * 100).round(2)
}).sort_values("Missing Percentage", ascending=False)

display(missing)
```

```python
import pandas as pd
import os

file_summary = []

for file, part in zip(files, dataframes):
    dates = pd.to_datetime(part["Date received"], errors="coerce")

    file_summary.append({
        "File": os.path.basename(file),
        "Rows": len(part),
        "Invalid dates": dates.isna().sum(),
        "Earliest date": dates.min(),
        "Latest date": dates.max(),
        "2022 complaints": (dates.dt.year == 2022).sum(),
        "2026 complaints": (dates.dt.year == 2026).sum(),
        "Unique complaint IDs": part["Complaint ID"].nunique()
    })

file_summary_df = pd.DataFrame(file_summary)

display(file_summary_df)
```

```python
import pandas as pd
import os

monthly_checks = []

for file, part in zip(files, dataframes):
    dates = pd.to_datetime(part["Date received"], errors="coerce")

    counts = (
        dates.dropna()
        .dt.to_period("M")
        .value_counts()
        .sort_index()
    )

    for month, count in counts.items():
        monthly_checks.append({
            "Source file": os.path.basename(file),
            "Month": str(month),
            "Complaint count": int(count)
        })

monthly_check_df = pd.DataFrame(monthly_checks)

# Display the month-by-month counts for each source file
for filename, group in monthly_check_df.groupby("Source file", sort=False):
    print(f"\n--- {filename} ---")
    print(group[["Month", "Complaint count"]].to_string(index=False))
```
